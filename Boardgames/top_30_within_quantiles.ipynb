```python
import pandas as pd

# Load data
games = pd.read_csv("games.csv")

# Calculate quantile boundaries based on AvgRating
q25 = games["AvgRating"].quantile(0.25)
q50 = games["AvgRating"].quantile(0.50)
q75 = games["AvgRating"].quantile(0.75)

# Divide games into quantile groups
groups = {
    "0–25% (Lowest ratings)": games[games["AvgRating"] <= q25],
    "25–50%": games[(games["AvgRating"] > q25) & (games["AvgRating"] <= q50)],
    "50–75%": games[(games["AvgRating"] > q50) & (games["AvgRating"] <= q75)],
    "75–100% (Highest ratings)": games[games["AvgRating"] > q75]
}

# Display first 30 games in each group
for name, group in groups.items():
    print(f"\n{'='*60}")
    print(f"{name} (n={len(group)})")
    print("="*60)

    print(
        group[["Name", "AvgRating"]]
        .head(30)
        .to_string(index=False)
    )
```
